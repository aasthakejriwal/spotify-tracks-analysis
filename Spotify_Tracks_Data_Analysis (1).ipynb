```python
print("INFORMATION ABOUT THE DATASET")
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
print("Cleaning of the Dataset")
df = df.copy()

df = df.drop_duplicates()

text_columns = ['artists', 'album_name', 'track_name']

for col in text_columns:
    df[col] = df[col].fillna('Unknown')

print("Dataset cleaned successfully!")
print("Updated dataset shape:", df.shape)
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
print("Graph 1: Distribution of Track Popularity")
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='popularity', bins=20, kde=True)

plt.title('Distribution of Spotify Track Popularity')
plt.xlabel('Popularity Score')
plt.ylabel('Number of Tracks')

plt.show()
```

```python

print("Minimum popularity:", df['popularity'].min())
print("Maximum popularity:", df['popularity'].max())
print("Average popularity:", round(df['popularity'].mean(), 2))
print("Median popularity:", df['popularity'].median())
```

```python
print("Graph 2: Top 10 Most Popular Tracks")
import matplotlib.pyplot as plt
import seaborn as sns

# Select the 10 most popular tracks
top_tracks = df.nlargest(10, 'popularity').copy()

# Combine track and artist names for labels
top_tracks['track_label'] = (
    top_tracks['track_name'].fillna('Unknown')
    + ' - '
    + top_tracks['artists'].fillna('Unknown')
)

# Create the bar chart
plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_tracks,
    x='popularity',
    y='track_label'
)

plt.title('Top 10 Most Popular Spotify Tracks')
plt.xlabel('Popularity Score')
plt.ylabel('Track - Artist')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
print("Graph 3: Average Popularity by Genre ")
top_tracks = df.nlargest(10, 'popularity').copy()
top_tracks['label'] = top_tracks['track_name'].fillna('Unknown')

plt.figure(figsize=(12, 8))
sns.barplot(data=top_tracks, x='popularity', y='label', hue='label', legend=False)

plt.title('Top 10 Most Popular Spotify Tracks')
plt.xlabel('Popularity Score')
plt.ylabel('Track Name')
plt.tight_layout()
plt.show()
```

```python
print("Graph 4: Danceability vs Popularity ")
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average popularity for each genre
genre_popularity = (
    df.groupby('track_genre')['popularity']
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

# Create the bar chart
plt.figure(figsize=(12, 6))

sns.barplot(
    data=genre_popularity,
    x='popularity',
    y='track_genre'
)

plt.title('Top 10 Genres by Average Track Popularity')
plt.xlabel('Average Popularity Score')
plt.ylabel('Music Genre')
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
print("Graph 5: Energy vs Valence ")
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='danceability',
    y='popularity',
    alpha=0.3
)

plt.title('Danceability vs Spotify Track Popularity')
plt.xlabel('Danceability')
plt.ylabel('Popularity Score')

plt.tight_layout()
plt.show()
```

```python

correlation = df['danceability'].corr(df['popularity'])

print("Correlation between danceability and popularity:",
      round(correlation, 3))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='energy',
    y='valence',
    alpha=0.3
)

plt.title('Energy vs Valence of Spotify Tracks')
plt.xlabel('Energy')
plt.ylabel('Valence')

plt.tight_layout()
plt.show()
```

```python

correlation = df['energy'].corr(df['valence'])

print("Correlation between energy and valence:",
      round(correlation, 3))
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
print("Graph 6: Audio Feature Correlation Heatmap")
audio_features = [
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'instrumentalness',
    'liveness',
    'valence',
    'tempo',
    'popularity'
]

# Calculate correlations
correlation_matrix = df[audio_features].corr()

# Plot the heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Spotify Audio Features')
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
print("Graph 7: Track Duration Distribution ")
df['duration_min'] = df['duration_ms'] / 60000

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df[df['duration_min'] <= 15],
    x='duration_min',
    bins=30,
    kde=True
)

plt.title('Distribution of Spotify Track Duration')
plt.xlabel('Track Duration (Minutes)')
plt.ylabel('Number of Tracks')

plt.tight_layout()
plt.show()
```

```python

print("Average duration (minutes):",
      round(df['duration_min'].mean(), 2))

print("Median duration (minutes):",
      round(df['duration_min'].median(), 2))

print("Shortest track (minutes):",
      round(df['duration_min'].min(), 2))

print("Longest track (minutes):",
      round(df['duration_min'].max(), 2))

```

```python

import matplotlib.pyplot as plt
import seaborn as sns

print("Graph 8: Explicit vs Non-Explicit Tracks")
explicit_counts = df['explicit'].map({
    True: 'Explicit',
    False: 'Non-Explicit'
}).value_counts()

# Create the bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=explicit_counts.index,
    y=explicit_counts.values
)

plt.title('Explicit vs Non-Explicit Spotify Tracks')
plt.xlabel('Track Type')
plt.ylabel('Number of Tracks')

plt.tight_layout()
plt.show()

```

```python

print("Track counts:")
print(df['explicit'].value_counts())

print("\nTrack percentages:")
print(
    (df['explicit'].value_counts(normalize=True) * 100)
    .round(2)
)
```
